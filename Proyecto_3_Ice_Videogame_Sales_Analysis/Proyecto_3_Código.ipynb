### 4.1 Top 5 Plataformas por Región

````python
def top5_platforms_by_region(df, region_col):
    sales = (
        df.groupby('platform')[region_col]
        .sum()
        .sort_values(ascending=False)
        .head(5)
    )
    
    share = (sales / df[region_col].sum() * 100).round(2)
    return pd.DataFrame({
        'sales': sales,
        'market_share_%': share
    })

top_na = top5_platforms_by_region(recent_data, 'na_sales')
top_eu = top5_platforms_by_region(recent_data, 'eu_sales')
top_jp = top5_platforms_by_region(recent_data, 'jp_sales')
````

## PASO 5: Pruebas de Hipótesis

Se utilizará un nivel de significancia α = 0.05.

Para comparar medias de dos grupos independientes se aplicará la prueba t de Student para muestras independientes (versión de Welch), ya que no se puede asumir igualdad de varianzas.

Si el p-value < α, se rechaza la hipótesis nula.
Si el p-value ≥ α, no se rechaza la hipótesis nula.

### Hipótesis 1: Xbox One vs PC (User Score)

H0: La media de `user_score` en Xbox One es igual a la media en PC.

H1: Las medias son diferentes.

````python
from scipy import stats

alpha = 0.05

xone = recent_data.loc[recent_data['platform'] == 'XOne', 'user_score'].dropna()
pc = recent_data.loc[recent_data['platform'] == 'PC', 'user_score'].dropna()

t_stat1, p_val1 = stats.ttest_ind(xone, pc, equal_var=False)

print("Hipótesis 1 - XOne vs PC")
print("t =", t_stat1)
print("p-value =", p_val1)

if p_val1 < alpha:
    print("Conclusión: Rechazamos H0 (hay diferencia significativa).")
else:
    print("Conclusión: No rechazamos H0 (no hay evidencia suficiente de diferencia).")
````

### Hipótesis 2: Action vs Sports (User Score)

H0: La media de `user_score` en Action es igual a la media en Sports.

H1: Las medias son diferentes.

````python
action = recent_data.loc[recent_data['genre'] == 'Action', 'user_score'].dropna()
sports = recent_data.loc[recent_data['genre'] == 'Sports', 'user_score'].dropna()

t_stat2, p_val2 = stats.ttest_ind(action, sports, equal_var=False)

print("Hipótesis 2 - Action vs Sports")
print("t =", t_stat2)
print("p-value =", p_val2)

if p_val2 < alpha:
    print("Conclusión: Rechazamos H0 (hay diferencia significativa).")
else:
    print("Conclusión: No rechazamos H0 (no hay evidencia suficiente de diferencia).")
````

### Conclusión de las pruebas de hipótesis

En la comparación entre Xbox One y PC, no se encontró evidencia estadística suficiente para afirmar que las calificaciones promedio de los usuarios sean diferentes (p > 0.05). Esto sugiere que, en términos de percepción del usuario, ambas plataformas presentan niveles similares de satisfacción.

En cambio, al comparar los géneros Action y Sports, se encontró una diferencia estadísticamente significativa en las calificaciones promedio (p < 0.05). Esto indica que los usuarios perciben de manera distinta estos dos géneros.

En conclusión, mientras que la plataforma no parece influir significativamente en la calificación promedio del usuario, el género sí representa un factor relevante en la percepción del público.

En primer lugar, se observó que la industria presenta ciclos claros de vida para las plataformas, con una duración aproximada de 7 a 10 años. Durante el periodo reciente (2013–2016), las plataformas líderes fueron PS4 y Xbox One, mientras que generaciones anteriores como PS3 y X360 mostraron señales de declive. Esto indica que las campañas publicitarias para 2017 deberían enfocarse principalmente en consolas de nueva generación.

En cuanto a la influencia de las reseñas, se encontró una correlación positiva moderada entre las puntuaciones de críticos y las ventas, lo que sugiere que la evaluación profesional impacta en el desempeño comercial. Sin embargo, las calificaciones de usuarios no mostraron una relación significativa con las ventas, indicando que otros factores como marketing, franquicias reconocidas y popularidad de género también desempeñan un papel importante.

El análisis por género reveló que Action y Shooter dominan en Norteamérica y Europa, mientras que Role-Playing lidera claramente en Japón. Estas diferencias culturales deben considerarse en la planificación de lanzamientos y estrategias de marketing regional.

Asimismo, el rating ESRB mostró impacto en mercados occidentales, donde los juegos clasificados como M (Mature) generan mayores ventas, mientras que en Japón las diferencias por clasificación son menos marcadas.

Las pruebas de hipótesis confirmaron que no existen diferencias significativas en la percepción del usuario entre Xbox One y PC, pero sí existen diferencias claras entre géneros como Action y Sports.

En conclusión, el éxito comercial de un videojuego depende de una combinación de factores: plataforma adecuada, género alineado con preferencias regionales y, en menor medida, evaluación crítica. Para maximizar el rendimiento en 2017, la tienda Ice debería priorizar títulos de acción y disparos en PS4 y Xbox One para mercados occidentales, y títulos RPG en plataformas dominantes en Japón.
