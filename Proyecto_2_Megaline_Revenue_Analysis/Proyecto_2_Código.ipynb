## Agregar datos por usuario

[Ahora que los datos están limpios, agrega los datos por usuario y por periodo para que solo haya un registro por usuario y por periodo. Esto facilitará mucho el análisis posterior.]

```python
calls_month = calls.groupby(['user_id', 'month']).agg(
    calls_count=('id', 'count'),
    minutes_used=('duration', 'sum')
).reset_index()

calls_month.head()
```

```python
minutes_per_month = calls.groupby(['user_id', 'month']).agg(
    minutes_used=('duration', 'sum')
).reset_index()

minutes_per_month.head()
```

```python
messages_month = messages.groupby(['user_id', 'month']).agg(
    messages_count=('id', 'count')
).reset_index()

messages_month.head()
```

```python
internet_month = internet.groupby(['user_id', 'month']).agg(
    mb_used=('mb_used', 'sum')
).reset_index()
internet_month['gb_used'] = np.ceil(internet_month['mb_used'] / 1024)

internet_month.head()

```

[Junta los datos agregados en un DataFrame para que haya un registro que represente lo que consumió un usuario único en un mes determinado.]

```python
usage = calls_month.merge(messages_month, 
                          on=['user_id', 'month'], 
                          how='outer') \
                   .merge(internet_month[['user_id', 'month', 'gb_used']], 
                          on=['user_id', 'month'], 
                          how='outer') \
                   .fillna(0)

usage.head()
```

```python
usage = usage.merge(users[['user_id', 'plan', 'city']], 
                    on='user_id', 
                    how='left')
usage = usage.merge(plans, 
                    left_on='plan', 
                    right_on='plan_name', 
                    how='left')

usage.head()

```

[Calcula los ingresos mensuales por usuario (resta el límite del paquete gratuito del número total de llamadas, mensajes de texto y datos; multiplica el resultado por el valor del plan de llamadas; añade la tarifa mensual en función del plan de llamadas). Nota: Dadas las condiciones del plan, ¡esto podría no ser tan trivial como un par de líneas! Así que no pasa nada si dedicas algo de tiempo a ello.]

```python
usage['gb_included'] = (usage['mb_per_month_included'] // 1024).astype(int)
usage['extra_minutes'] = np.maximum(
    usage['minutes_used'] - usage['minutes_included'], 0
)

usage['extra_messages'] = np.maximum(
    usage['messages_count'] - usage['messages_included'], 0
)

usage['extra_gb'] = np.maximum(
    usage['gb_used'] - usage['gb_included'], 0
)

usage['monthly_revenue'] = (
    usage['usd_monthly_pay']  # ← Aquí está el cambio
    + usage['extra_minutes'] * usage['usd_per_minute']
    + usage['extra_messages'] * usage['usd_per_message']
    + usage['extra_gb'] * usage['usd_per_gb']
)

usage[['user_id', 'month', 'plan', 'monthly_revenue']].head()
```

El cálculo del exceso de internet debe coincidir con la regla de facturación con total precisión: el total mensual se redondea hacia arriba a GB enteros, y la franquicia incluida debe compararse en esos mismos términos de GB enteros para que el exceso sea un número entero de GB. Ahora mismo estás mezclando una franquicia flotante (`gb_included`) con el uso redondeado, lo que puede crear inconsistencias sutiles—especialmente si la cantidad incluida no se representa como un valor de GB entero.

## Prueba las hipótesis estadísticas

[Prueba la hipótesis de que son diferentes los ingresos promedio procedentes de los usuarios de los planes de llamada Ultimate y Surf.]

[Elabora las hipótesis nula y alternativa, escoge la prueba estadística, determina el valor alfa.]

```python
surf_income = usage[usage['plan'] == 'surf']['monthly_revenue']
ultimate_income = usage[usage['plan'] == 'ultimate']['monthly_revenue']
alpha = 0.05

t_stat, p_value = stats.ttest_ind(
    surf_income,
    ultimate_income,
    equal_var=False
)

print("Estadístico t:", t_stat)
print("Valor p:", p_value)

if p_value < alpha:
    print("Se rechaza la hipótesis nula (H0).")
else:
    print("No se rechaza la hipótesis nula (H0).")
```

Prueba de hipótesis: comparación de ingresos entre Surf y Ultimate
1. Planteamiento del problema

Se desea determinar si el ingreso promedio mensual generado por los usuarios del plan Surf es diferente al del plan Ultimate.

2. Hipótesis

Hipótesis nula (H₀):
El ingreso promedio mensual es igual en ambos planes.

𝜇𝑠𝑢𝑟𝑓 = 𝜇𝑢𝑙𝑡𝑖𝑚𝑎𝑡𝑒μ 

Hipótesis alternativa (H₁):
El ingreso promedio mensual es diferente entre los planes.

𝜇𝑠𝑢𝑟𝑓≠ 𝜇𝑢𝑙𝑡𝑖𝑚𝑎𝑡𝑒μsurf

Se trata de una prueba bilateral.

3. Prueba estadística seleccionada

Se utilizó una prueba t de Student para muestras independientes (Welch’s t-test) porque:
Se comparan medias de dos grupos independientes.
No se asume igualdad de varianzas.
El tamaño de muestra es suficientemente grande.

4. Nivel de significancia

Se estableció un nivel de significancia:
α=0.05

5. Resultados

Estadístico t = −7.95

Valor p = 3.17 × 10⁻¹⁵

Dado que el valor p es mucho menor que 0.05, se rechaza la hipótesis nula.

6. Conclusión

Existe evidencia estadística sólida para concluir que el ingreso promedio mensual difiere significativamente entre los planes Surf y Ultimate.

La diferencia observada no se debe al azar, sino que refleja un comportamiento distinto en los ingresos generados por cada plan.
